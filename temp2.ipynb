Group Bar chart

```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
boys_marks = [85, 90, 78, 92, 88, 95]
girls_marks = [80, 85, 75, 90, 82, 88]
hours = [1, 2, 3, 4, 5, 6]
plt.bar(hours, boys_marks, color='blue', label='Boys')
plt.bar(hours, girls_marks, color='red', label='Girls')
plt.title('Study Hours vs Marks')
plt.xlabel('Hours Studied')
plt.ylabel('Marks')
plt.legend()
```
