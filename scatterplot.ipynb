Scatter Diagram

```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
hours = [1, 2, 3, 4, 5, 6]
boys_marks = [85, 90, 78, 92, 88, 95]
girls_marks = [80, 85, 75, 90, 82, 88]
plt.scatter(hours, boys_marks, color='blue', label='Boys')
plt.scatter(hours, girls_marks, color='red', label='Girls')
plt.title('Study Hours vs Marks')
plt.xlabel('Hours Studied')
plt.ylabel('Marks')
plt.legend()
plt.plot(hours, boys_marks, label='Trend Line (Boys)', color='blue')
plt.plot(hours, girls_marks, label='Trend Line (Girls)', color='red')
plt.show()
```
