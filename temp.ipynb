```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
# colors = random_colors = np.random.rand(len(students), 3)
colors = ['red', 'blue', 'green', 'orange', 'purple']
students = ['Alice', 'Bob', 'Charlie', 'David', 'Eve']
marks = [85, 90, 78, 92, 88]
```

```python
plt.bar(students, marks, color=colors)
plt.title('Student Marks')
plt.xlabel('Students')
plt.ylabel('Marks')
plt.ylim(0, 100)
for i in range(len(students)):
    plt.text(i, marks[i], str(marks[i]), ha='center', va='bottom')
plt.show()
```

```python
time = [11, 12, 1, 2, 3]
pizza = [50, 60, 45, 70, 65]
italian = [40, 55, 35, 60, 50]
chinese = [20, 30, 15, 40, 25]
punjabi = [30, 45, 25, 50, 40]
# colors = random_colors = np.random.rand(4, 4)
colors = ['red', 'blue', 'green', 'orange']

plt.bar(time, pizza, color=colors[0])
plt.bar(time, pizza, color=colors[1], bottom=italian)
plt.bar(time, pizza, color=colors[2], bottom = [italian[i] + punjabi[i] for i in range(len(time))])
plt.bar(time, pizza, color=colors[3], bottom = [italian[i] + punjabi[i] + chinese[i] for i in range(len(time))])
plt.xlim(1, 12)
plt.ylim(0, 300)
plt.xlabel('Time (Hours)')
plt.ylabel('Number of Orders')
plt.title('Food Orders Over Time')
plt.legend(['Pizza', 'Italian', 'Punjabi', 'Chinese'])
plt.xticks(time)
for i in range(len(time)):
    plt.text(time[i], pizza[i]/2, str(pizza[i]), ha='center', va='bottom', color='white')
    plt.text(time[i], pizza[i] + italian[i]/2, str(italian[i]), ha='center', va='bottom', color='white')
    plt.text(time[i], pizza[i] + italian[i] + punjabi[i]/2, str(punjabi[i]), ha='center', va='bottom', color='white')
    plt.text(time[i], pizza[i] + italian[i] + punjabi[i] + chinese[i]/2, str(chinese[i]), ha='center', va='bottom', color='white')
plt.show()
```
