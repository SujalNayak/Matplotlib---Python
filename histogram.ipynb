```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
```

```python
#50000 < salary < 150000
salary = np.random.randint(50000,150000,1000)
plt.hist(salary, bins=50)
plt.xlabel('Salary')
plt.ylabel('Frequency')
plt.title('Salary Distribution')
```

```python
boys = np.random.randint(10, 100, 20) 
girls = np.random.randint(10, 100, 20)
plt.boxplot([boys, girls], vert=False)
plt.xticks([1, 2], ['Boys', 'Girls'])
plt.title('Box Plot of Boys and Girls')
```

```python
ticket_Price = np.random.randint(100, 1000, 10)
destinations = ['Ahmedabad', 'Mumbai', 'Delhi', 'Bangalore', 'Chennai', 'Kolkata', 'Hyderabad', 'Pune', 'Jaipur', 'Lucknow']
plt.bar(destinations, ticket_Price)
plt.xticks(rotation=45)
plt.title('Ticket Prices for Destinations')
plt.xlabel('Destinations')
plt.ylabel('Ticket Price (INR)')
```

```python
fig,axs = plt.subplots(2,2, figsize=(10,10))
axs[0,0].hist(ticket_Price, bins=10)
axs[0,0].set_title('Histogram of Ticket Prices')
axs[0,0].set_xlabel('Ticket Price (INR)')
axs[0,0].set_ylabel('Frequency')
axs[0,1].bar(destinations, ticket_Price)
axs[0,1].set_title('Bar Chart of Ticket Prices')
axs[0,1].set_xlabel('Destinations')
axs[0,1].set_ylabel('Ticket Price (INR)')
axs[0,1].tick_params(axis='x', rotation=45)
axs[1,0].boxplot(ticket_Price, vert=False)
axs[1,0].set_title('Box Plot of Ticket Prices')
axs[1,0].set_ylabel('Ticket Price (INR)')
axs[1,1].scatter(destinations, ticket_Price)
axs[1,1].set_title('Scatter Plot of Ticket Prices')
axs[1,1].set_xlabel('Destinations')
axs[1,1].set_ylabel('Ticket Price (INR)')
axs[1,1].tick_params(axis='x', rotation=45)
```

```python
days = ['Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday', 'Saturday', 'Sunday']
traffic_Volume = np.random.randint(50, 500, (7, 24))
sns.heatmap(traffic_Volume, cmap='YlOrRd', yticklabels=days, xticklabels=range(24))
plt.title('Traffic Volume Heatmap')
plt.xlabel('Hours')
plt.ylabel('Days of the Week')
```
